```python
import matplotlib.pyplot as plt
import seaborn as sns
```

### Univariate Analysis of Numerical Columns

First, let's identify the numerical and categorical columns. Then, we'll examine the descriptive statistics and distributions of the numerical columns.

```python
numerical_cols = df.select_dtypes(include=np.number).columns.tolist()
categorical_cols = df.select_dtypes(include='object').columns.tolist()

print("Numerical columns:", numerical_cols)
print("Categorical columns:", categorical_cols)

# Display descriptive statistics for numerical columns
display(df[numerical_cols].describe())
```

Now, let's visualize the distributions of the numerical columns using histograms and box plots. Histograms show the frequency distribution, while box plots help identify outliers and understand the spread.

```python
for col in numerical_cols:
    plt.figure(figsize=(12, 5))

    # Histogram
    plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
    sns.histplot(df[col], kde=True)
    plt.title(f'Distribution of {col}')
    plt.xlabel(col)
    plt.ylabel('Frequency')

    # Box Plot
    plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
    sns.boxplot(y=df[col])
    plt.title(f'Box Plot of {col}')
    plt.ylabel(col)

    plt.tight_layout()
    plt.show()
```

### Bivariate Analysis

Now, let's explore the relationships between different pairs of variables using bivariate analysis. This can reveal patterns, correlations, and dependencies that univariate analysis might miss.

#### Numerical vs. Numerical: Scatter Plots and Correlation

We'll use scatter plots to visualize the relationship between two numerical variables and calculate their correlation coefficient to quantify the strength and direction of their linear relationship.

```python
# Relationship between Booking Value and Ride Distance
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Ride Distance', y='Booking Value', data=df)
plt.title('Booking Value vs. Ride Distance')
plt.xlabel('Ride Distance (km)')
plt.ylabel('Booking Value')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print(f"Correlation between Booking Value and Ride Distance: {df['Booking Value'].corr(df['Ride Distance']):.2f}")

# Relationship between Driver Ratings and Customer Rating
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Driver Ratings', y='Customer Rating', data=df)
plt.title('Driver Ratings vs. Customer Rating')
plt.xlabel('Driver Ratings')
plt.ylabel('Customer Rating')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()

print(f"Correlation between Driver Ratings and Customer Rating: {df['Driver Ratings'].corr(df['Customer Rating']):.2f}")
```

#### Categorical vs. Numerical: Box Plots and Violin Plots

Box plots and violin plots are excellent for comparing the distribution of a numerical variable across different categories of a categorical variable. This helps us see how the numerical values vary within each group.

```python
# Booking Value by Booking Status
plt.figure(figsize=(12, 7))
sns.boxplot(x='Booking Status', y='Booking Value', data=df)
plt.title('Booking Value by Booking Status')
plt.xlabel('Booking Status')
plt.ylabel('Booking Value')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()

# Ride Distance by Vehicle Type
plt.figure(figsize=(12, 7))
sns.violinplot(x='Vehicle Type', y='Ride Distance', data=df)
plt.title('Ride Distance by Vehicle Type')
plt.xlabel('Vehicle Type')
plt.ylabel('Ride Distance (km)')
plt.xticks(rotation=45, ha='right')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

#### Categorical vs. Categorical: Count Plots and Heatmaps (Cross-tabulation)

For two categorical variables, count plots can show the frequency of each combination, while heatmaps of cross-tabulations (contingency tables) provide a clear visual representation of their joint distribution.

```python
# Count of Booking Status by Vehicle Type
plt.figure(figsize=(14, 8))
sns.countplot(x='Booking Status', hue='Vehicle Type', data=df, palette='viridis')
plt.title('Booking Status Distribution by Vehicle Type')
plt.xlabel('Booking Status')
plt.ylabel('Count')
plt.xticks(rotation=45, ha='right')
plt.legend(title='Vehicle Type')
plt.tight_layout()
plt.show()

# Heatmap of Booking Status vs. Payment Method
crosstab_df = pd.crosstab(df['Booking Status'], df['Payment Method'])
plt.figure(figsize=(12, 8))
sns.heatmap(crosstab_df, annot=True, fmt='d', cmap='YlGnBu', linewidths=.5)
plt.title('Booking Status vs. Payment Method (Count)')
plt.xlabel('Payment Method')
plt.ylabel('Booking Status')
plt.tight_layout()
plt.show()
```

### Multivariate Analysis

Multivariate analysis helps us understand relationships between three or more variables simultaneously. A common approach is to look at the correlation between all numerical features.

#### Correlation Heatmap of Numerical Features

We can calculate the correlation matrix for all numerical columns and visualize it as a heatmap. This provides a quick overview of how each numerical variable relates to every other numerical variable, highlighting strong positive or negative correlations.

```python
plt.figure(figsize=(12, 10))
sns.heatmap(df[numerical_cols].corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
plt.title('Correlation Matrix of Numerical Features')
plt.tight_layout()
plt.show()
```

#### Pair Plot for Selected Numerical Features

A pair plot creates a grid of scatter plots for each pair of variables and histograms for individual variables. This can be very useful for spotting trends, relationships, and distributions within subgroups.

```python
# Select a subset of numerical columns for a more readable pair plot
selected_numerical_cols = ['Booking Value', 'Ride Distance', 'Avg VTAT', 'Avg CTAT', 'Customer Rating']

sns.pairplot(df[selected_numerical_cols])
plt.suptitle('Pair Plot of Selected Numerical Features', y=1.02) # Adjust suptitle position
plt.tight_layout()
plt.show()
```

## Observations from Data Exploration

### Univariate Analysis Observations

Based on the histograms and box plots of numerical columns (`Avg VTAT`, `Avg CTAT`, `Booking Value`, `Ride Distance`, `Driver Ratings`, `Customer Rating`):

*   **Avg VTAT and Avg CTAT**: Appear to be generally right-skewed, indicating that most rides have lower average vehicle and customer arrival times, but there are some instances of significantly higher times, possibly due to traffic, long distances, or delays.
*   **Cancelled Rides by Customer/Driver, Incomplete Rides**: These columns were mostly filled with `0` (after imputation), suggesting that for completed rides or rides without these specific statuses, these metrics are not applicable or were zero. The original missing values suggest these are specific to non-completed statuses. When the values are `1.0`, it indicates a cancellation or incompleteness.
*   **Booking Value and Ride Distance**: Both show a wide range of values, with distributions that are generally right-skewed, meaning many shorter, lower-value rides and fewer very long, high-value rides. Outliers in both suggest exceptional trips.
*   **Driver Ratings and Customer Rating**: Both tend to be high, clustered around 4.0-5.0, which is positive. The distributions appear somewhat left-skewed, indicating customer and driver satisfaction is generally good. There are some lower ratings which might be outliers indicating specific issues.

### Bivariate Analysis Observations

*   **Booking Value vs. Ride Distance**: The scatter plot showed a very weak positive correlation (`0.01`). This is unexpected; typically, longer rides would correlate with higher booking values. This weak correlation might suggest inconsistent pricing, a flat base fare structure, or other factors influencing pricing more significantly than distance.
*   **Driver Ratings vs. Customer Rating**: The correlation was near zero (`-0.00`). This indicates that a driver's rating does not linearly correlate with the customer's rating in this dataset. This could mean customer and driver satisfaction are independent metrics or influenced by different factors.
*   **Booking Value by Booking Status**: Box plots showed significant differences in `Booking Value` across `Booking Status` categories. For example, 'Completed' rides likely have the highest booking values, while 'No Driver Found', 'Cancelled by Customer', or 'Cancelled by Driver' would ideally have zero or very low booking values, which is consistent with the initial data cleaning where null `Booking Value` entries (likely for uncompleted rides) were imputed.
*   **Ride Distance by Vehicle Type**: Violin plots revealed distinct distributions of `Ride Distance` for different `Vehicle Type`s. For instance, 'Bike' or 'eBike' might be used for shorter distances, while 'Go Sedan' or 'Premier Sedan' for longer journeys, which is logical.
*   **Booking Status Distribution by Vehicle Type**: Count plots showed the frequency of different booking statuses for each vehicle type. For example, certain vehicle types might have a higher 'No Driver Found' rate, indicating supply issues.
*   **Booking Status vs. Payment Method (Heatmap)**: This showed the joint distribution. For example, 'Completed' rides are most frequent, and 'UPI' is a very popular payment method across all booking statuses where a payment method is recorded. It also highlights which payment methods are associated with cancellations.

### Multivariate Analysis Observations

*   **Correlation Matrix of Numerical Features**: The heatmap generally shows weak correlations among most numerical features. The strongest positive correlation is between `Booking Value` and `Ride Distance`, though still very low (as seen in bivariate analysis). `Avg VTAT` and `Avg CTAT` might show some positive correlation, as longer vehicle arrival times might lead to longer customer waiting times.
*   **Pair Plot of Selected Numerical Features**: Confirmed the distributions and weak pairwise relationships seen previously. It can help identify potential non-linear relationships or clusters if they existed, but in this case, it largely reinforced the observed weak correlations.

## Business Insights

Based on the observations, here are some key business insights:

1.  **Customer and Driver Satisfaction is Generally High**: `Driver Ratings` and `Customer Rating` are mostly favorable, indicating a generally positive experience for both drivers and customers.
2.  **Inefficiency in Pricing or Ride Matching**: The very weak correlation between `Booking Value` and `Ride Distance` suggests that pricing might not be optimally aligned with the actual distance traveled. This could lead to missed revenue opportunities for longer rides or overpriced shorter rides.
3.  **Significant Cancellation and Incompletion Rates**: While missing values for `Cancelled Rides` and `Incomplete Rides` were imputed, their original prevalence indicates these are significant issues. Understanding the primary reasons for cancellations (Customer vs. Driver) is crucial.
4.  **Payment Method Popularity**: UPI is a highly popular payment method. Understanding its usage across different ride statuses can help in optimizing payment systems and incentives.
5.  **Vehicle Type Utilization**: Different vehicle types are used for varying ride distances. This segmentation can be leveraged for targeted marketing and supply management.
6.  **Potential Service Bottlenecks**: The existence of 'No Driver Found' statuses, coupled with potentially high `Avg VTAT` and `Avg CTAT` outliers, suggests service bottlenecks in certain areas or during peak times, leading to unfulfilled demand.

## Recommendations

Here are some recommendations derived from the business insights:

1.  **Review Pricing Model**: Investigate the factors influencing `Booking Value` beyond `Ride Distance`. Consider adjusting the pricing algorithm to better reflect distance, time, and demand to optimize revenue and customer perception of fairness.
2.  **Deep Dive into Cancellations and Incomplete Rides**: Analyze the `Reason for cancelling by Customer` and `Driver Cancellation Reason` columns more thoroughly (after appropriate handling of `NaN` or 'Unknown' values) to identify common pain points. Implement strategies to mitigate these reasons, such as improved driver matching, clearer communication, or better vehicle maintenance.
3.  **Optimize Driver Supply and Demand**: For statuses like 'No Driver Found', analyze the `Pickup Location`, `Time`, and `Vehicle Type` to identify areas and times with insufficient driver supply. Implement dynamic pricing (surge pricing) or driver incentives to balance supply and demand.
4.  **Enhance Payment System**: Leverage the popularity of UPI by offering exclusive deals or loyalty programs for users paying via UPI. Explore reasons for cancellations associated with specific payment methods.
5.  **Targeted Vehicle Type Promotions**: Promote specific vehicle types for optimal use cases. For example, encourage eBike usage for short-distance commutes in congested areas, or offer incentives for sedan bookings for longer inter-city rides.
6.  **Improve ETA Prediction and Communication**: Analyze outliers in `Avg VTAT` and `Avg CTAT`. Improve real-time traffic prediction and communication to both drivers and customers to manage expectations and reduce frustration.
7.  **Customer and Driver Feedback Loop**: Implement robust mechanisms for collecting and acting on feedback, especially from those giving lower `Driver Ratings` or `Customer Rating`, to continuously improve service quality.
